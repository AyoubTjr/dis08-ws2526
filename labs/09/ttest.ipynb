```python
import pandas as pd
from scipy import stats
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(url)

df.head()
```

```python
df_clean = df[["sex", "body_mass_g"]].dropna()
df_clean.head()
```

```python
df_clean["sex"].value_counts()
```

```python
male_mass = df_clean[df_clean["sex"] == "MALE"]["body_mass_g"]
female_mass = df_clean[df_clean["sex"] == "FEMALE"]["body_mass_g"]

print("Male sample size:", len(male_mass))
print("Female sample size:", len(female_mass))
```

## Hypothesis

We test whether the mean body mass differs between male and female penguins.

- **Null hypothesis (H₀):**  
  The mean body mass of male and female penguins is equal.

- **Alternative hypothesis (H₁):**  
  The mean body mass of male and female penguins is different.

```python
t_stat, p_value = stats.ttest_ind(
    male_mass,
    female_mass,
    equal_var=False
)

t_stat, p_value
```

## Interpretation of Results

The p-value is below 0.05, indicating a statistically significant result
at the 5% significance level.

Therefore, we reject the null hypothesis.

This suggests that there is a significant difference in mean body mass
between male and female penguins.

## Conclusion

An independent two-sample t-test was conducted to compare body mass
between male and female penguins.

The test shows a statistically significant difference, suggesting that
sex is associated with differences in body mass in penguins.

This demonstrates how hypothesis testing can be applied to real-world
biological data.
